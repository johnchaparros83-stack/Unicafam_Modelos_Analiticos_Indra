# **Ejercicio 3**

Usando como referencia el ejercicio de ventas realizado en la clase

```python

import pandas as pd
import numpy as np
# Números aleatorios
np.random.seed(42)
n=200

# Creación de la tabla
# DataFrame= tablas con filas y columnas
ventas = pd.DataFrame ({
    "fecha": pd.date_range("2026-01-01",periods=n, freq="D"),
    "ciudad": np.random.choice(["Bogotá","Medellín","Cali","Barranquilla"], n),
    "canal": np.random.choice(["Tienda","Web","App"], n),
    "unidades": np.random.randint(1, 20, n),
    "precio": np.random.choice([25000, 48000, 99000], n),
})
# Creación de la columna ventas
# shape = filas, columnas
ventas["total"] = ventas["unidades"] * ventas["precio"]
print(ventas.shape)
print(ventas.head(3))
ventas.info()
```

```python
#Total vendido por canal
ventas.groupby("canal")["total"].sum()
```

```python
#Ticket promedio por ciudad
ventas.groupby("ciudad")["total"].mean()
```

```python
#Día de la semana en que se vende mas

#crear columna con el día de la semana
ventas["dia_semana"] = ventas["fecha"].dt.day_name()
#Agrupar por dia y sumar ventas
ventas.groupby("dia_semana")["total"].sum().sort_values(ascending=False)
#dia que más vende
dia=ventas.groupby("dia_semana")["total"].sum().idxmax()
print(dia)
```

Conclusión para el gerente: El análisis muestra que los viernes son los días de mayores ventas, se hacen principalmente por tiendas y Medellín tiene el ticket promedio más alto. Se recomienda mantener las estrategias exitosas, pero priorizar acciones de mejora en el canal Web y en ciudades como Cali y Bogotá, donde existen oportunidades para incrementar los ingresos y mejorar el desempeño comercial.
