# **Ejercicio 1**

```python
# Crear funcion para evaluar notas
#def para crear la función
#if condicional para clasificar
def evaluar_nota(nota):
    if nota < 0 or nota > 5:
        return "Nota inválida"
    elif nota < 3.0:
        return "Reprueba"
    elif nota < 4.5:
        return "Aprueba"
    else:
        return "Excelente"
```

```python
#Aplicar en lista de notas
notas = [1.7, 2.9, 3.4, 2.8, 5.0, 3.4]
# Inicializamos los aprobados en 0
aprobados = 0
#for para recorrer las notas
for nota in notas:
    estado = evaluar_nota(nota)
    print(nota, "->", estado)
# aumenta en 1 el contador para saber cuantos aprobaron
    if estado == "Aprueba" or estado == "Excelente":
        aprobados += 1

print("Total de estudiantes que aprobaron:", aprobados)
```

# **Ejercicio 2**

```python
# crear carrito precio, cantidad
carrito = {
    "camiseta": (46000, 5),
    "jean": (180000, 3),
    "medias": (24000, 7)
}
```

```python
# Recorrer el carrito
subtotal = 0
#.items(): recorrer las claves y los valores.
for producto, (precio, cantidad) in carrito.items():
    valor = precio * cantidad
    #Acumular
    subtotal += valor
    print(producto, ":", valor)

print("Subtotal:", subtotal)
```

```python
# Aplicar descuento 10%
if subtotal > 200000:
    descuento = subtotal * 0.10
else:
    descuento = 0

total_descuento = subtotal - descuento
#Verificar total y descuento
print ("Subtotal: ",subtotal, "Descuento: ", descuento, "Total con descuento: ",total_descuento)
```

```python
# Calcular Iva

iva = total_descuento * 0.19

total = total_descuento + iva
#f"${total:,.0f} para dejarlo en formato pesos
print("Subtotal:", f"${subtotal:,.0f}")
print("Descuento:", f"${descuento:,.0f}")
print("IVA (19%):", f"${iva:,.0f}")
print("TOTAL A PAGAR:", f"${total:,.0f}")
```

# **Ejercicio 3**

```python
# creé un DataFrame de empleados
import pandas as pd
import numpy as np

empleados = pd.DataFrame({
    "nombre": ["Camila", "Pedro", "Juan"],
    "area": ["Ventas", "Sistemas", "Recursos Humanos"],
    "salario": [3700000, 4800000, 3450000]
})

print(empleados)
```

```python

nuevos = pd.DataFrame({
    "nombre": ["John", "Carla", "Maria"],
    "area": ["Ventas", "Nomina", "Servicio al cliente"],
    "salario": [3700000, 4100000, 2000000]
})

#Agregar usuarios usando pd.concat
empleados = pd.concat([empleados, nuevos], ignore_index=True)

print(empleados)
```

```python
#Crear columna bono usando np.where para validar el area y el salario
empleados["bono"] = np.where(
    empleados["area"] == "Ventas",
    empleados["salario"] * 0.10,
    empleados["salario"] * 0.05
)

print(empleados)
```

```python
#Calcular costo nominal por área
empleados["total_nomina"] = empleados["salario"] + empleados["bono"]

nomina = empleados.groupby("area")["total_nomina"].sum()

print(nomina)
```

```python
#Empleado con mayor salario
#.idmax para encontrar el maximo
mayor_salario = empleados.loc[empleados["salario"].idxmax()]

print("El empleado con mayor salario es ", mayor_salario["nombre"])
```
