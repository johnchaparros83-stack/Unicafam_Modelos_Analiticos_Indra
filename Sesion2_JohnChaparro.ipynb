# **Taller Big Data**

```python
# Instalación de bibliotecas
import pandas as pd
import numpy as np
```

```python
# Números aleatorios
np.random.seed(42)
n=200
```

```python
# Creación de la tabla
# DataFrame= tablas con filas y columnas
ventas = pd.DataFrame ({
    "fecha": pd.date_range("2026-01-01",periods=n, freq="D"),
    "ciudad": np.random.choice(["Bogotá","Medellín","Cali","Barranquilla"], n),
    "canal": np.random.choice(["Tienda","Web","App"], n),
    "unidades": np.random.randint(1, 20, n),
    "precio": np.random.choice([25000, 48000, 99000], n),
})
```

```python
# Creación de la columna ventas
# shape = filas, columnas
ventas["total"] = ventas["unidades"] * ventas["precio"]
```

```python
#visualizar los primeros 3 resultados
print(ventas.shape)
print(ventas.head(3))
```

```python
ventas.info()
```

**ANALÍTICA DESCRIPTIVA: ¿QUÉ PASÓ?**

```python
# Tabla dinamica
resumen = (ventas.groupby ("ciudad")["total"]
.agg(["count", "sum", "mean"])
.sort_values("sum", ascending=False))
```

```python
# Resumen de los datos
#Convertir los valores a millones
resumen["sum"] = (resumen["sum"] / 1e6).round(1)
#Redondear el promedio
resumen["mean"] = resumen["mean"].round(0)
#Cambiar el nombre de las columnas
resumen.columns = ["ventas", "total_millones", "ticket_prom"]
print(resumen)
```

```python
# Crear tabla dinamica
tabla = pd.pivot_table(
    ventas,
    values="total",
    index="ciudad",
    columns="canal",
    aggfunc="sum"
) / 1e6
```

```python
# Mostrar resultados
print(tabla.round(1))
print()
print("Canal con más ventas:", ventas.groupby("canal")["total"].sum().idxmax())
```
