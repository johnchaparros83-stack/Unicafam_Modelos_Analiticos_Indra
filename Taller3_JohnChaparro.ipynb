# **3. ¿Te alcanza para el café?**

```python
presupuesto = 220000
precio_cafe = 4900
cafes_semana = 7

#1. ¿Cuánto gastas al mes? (4 semanas)
gasto_mes = precio_cafe*cafes_semana*4
print ("Al mes me gasto", gasto_mes)

#2 ¿Cuanto te sobra?
#abs para devolver valor absoluto
sobra = presupuesto - gasto_mes
if sobra > 0:
    print("Me sobra:", sobra)
else:
    print("Me falta:", abs(sobra))

#3. ¿Te alcanza? (True o False)
if gasto_mes <= presupuesto:
    print("Alcanza= ",True)
else:
    print("Alcanza ",False)
```

# **8. tu meta de ahorro**

```python
ahorro = 0        # ← lo que ya tienes
meta = 5000000       # ← cambia
por_mes = 750000    # ← cambia
meses = 0

# Mientras no llegues a la meta:
# suma lo del mes y cuenta un mes más
# Sale del while hasta que el ahorro supera la meta y se va guardando en la variable meses el mes en el que va el ahorro
while ahorro < meta:
    ahorro = ahorro + por_mes
    meses = meses + 1
    print("Mes", meses, "- Ahorro:", ahorro)
# ¿Cuántos meses tardaste?
print("Llegaste a la meta en", meses, "meses")
```

# **9. Dividir la cuenta**

```python
cuenta = 152000    # ← cambia
propina = 10     # ← cambia (%)
amigos = 7       # ← cambia

#Se crean las funciones para poder usarlas las veces que sea necesario
# 1. Regresa la cuenta más la propina
def con_propina(cuenta, porcentaje):
    return cuenta + (cuenta * porcentaje / 100)

# 2. Regresa cuánto paga cada persona
def por_persona(total, personas):
    return total / personas

# 3. Úsalas con tus datos e imprime
total = con_propina(cuenta, propina)
pago = por_persona(total, amigos)

print("Cuenta con propina:", total)
print("Pago por persona:", pago)

#Crear funcion nueva con descuento
def descuento(total, porcentaje):
    return total - (total * porcentaje / 100)

print("Cuenta con descuento del 10%:", descuento(total, 10))

### Dar formato $
print ("Con formato $")
print("Cuenta con propina: ${:,.0f}".format(total))
print("Pago por persona: ${:,.2f}".format(pago))
print("Cuenta con descuento del 10%: ${:,.0f}".format(descuento(total, 10)))
```

# **10. La caja de la tienda**

```python
# ← cambia por 4 cosas que comprarías
precios = [500,900,250,75]

# 1. Suma el total con un for
total = 0

for precio in precios:
    total += precio

# 2. Si el total pasa de 500,
#    aplica 10 % de descuento
#validar si supera 500 para aplicar el descuento
if total > 500:
    descuento = total * 0.10
else:
    descuento = 0

a_pagar = total - descuento


# 3. Imprime el ticket con f-strings:
print(f"Total: ${total:.2f}")
print(f"Descuento: ${descuento:.2f}")
print(f"A pagar: ${a_pagar:.2f}")
```
